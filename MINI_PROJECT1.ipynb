```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Dataset
df = pd.read_excel("TN_Education_Dataset.xlsx")

# 2. Overall Preference Totals

totals = df[["Bachelor","Engineering","Integrated"]].sum()

#  Bar Chart
plt.figure()
plt.bar(totals.index, totals.values)
plt.title("Overall Higher Education Preference – Tamil Nadu")
plt.ylabel("Number of Students")
plt.show()
```

```python
#  Pie Chart
#plt.figure()
plt.pie(totals, labels=totals.index, autopct="%1.1f%%", startangle=90)
plt.title("Overall Preference Distribution")
plt.show()


```

```python
# 3. District-wise Engineering vs Bachelor

plt.figure(figsize=(10,6))
plt.plot(df["District"], df["Engineering"], label="Engineering")
plt.plot(df["District"], df["Bachelor"], label="Bachelor")
plt.xticks(rotation=90)
plt.legend()
plt.title("Engineering vs Bachelor – District Comparison")
plt.ylabel("Number of Students")
plt.show()


```

```python
# 4. Stacked Bar Chart (All Courses)

plt.figure(figsize=(10,6))
plt.bar(df["District"], df["Bachelor"], label="Bachelor")
plt.bar(df["District"], df["Engineering"], bottom=df["Bachelor"], label="Engineering")
plt.bar(
    df["District"],
    df["Integrated"],
    bottom=df["Bachelor"] + df["Engineering"],
    label="Integrated"
)
plt.xticks(rotation=90)
plt.legend()
plt.title("Stacked View of Higher Education Choices by District")
plt.ylabel("Number of Students")
plt.show()


```

```python
# 5. Interest Gap Analysis

df["Eng_Bach_Gap"] = df["Engineering"] - df["Bachelor"]

plt.figure()
plt.bar(df["District"], df["Eng_Bach_Gap"])
plt.xticks(rotation=90)
plt.title("Engineering vs Bachelor Interest Gap")
plt.ylabel("Student Difference")
plt.show()


```

```python
# 6. Pass Rate vs Engineering Interest

df["Engineering %"] = (df["Engineering"] / df["Passed Students"]) * 100

plt.figure()
plt.scatter(df["Pass %"], df["Engineering %"])
plt.xlabel("Pass Percentage")
plt.ylabel("Engineering Interest (%)")
plt.title("Pass Rate vs Engineering Interest")
plt.show()
```

```python


# 7. Top & Bottom District Comparison

top5 = df.sort_values("Engineering %", ascending=False).head(5)
bottom5 = df.sort_values("Engineering %").head(5)

plt.figure()
plt.bar(top5["District"], top5["Engineering %"])
plt.title("Top 5 Districts – Engineering Interest")
plt.ylabel("Engineering %")
plt.show()

plt.figure()
plt.bar(bottom5["District"], bottom5["Engineering %"])
plt.title("Bottom 5 Districts – Engineering Interest")
plt.ylabel("Engineering %")
plt.show()


```

```python
# 8. District Contribution Share

df["Engineering Share %"] = (df["Engineering"] / totals["Engineering"]) * 100

top_contributors = df.sort_values("Engineering Share %", ascending=False).head(7)

plt.figure()
plt.pie(
    top_contributors["Engineering Share %"],
    labels=top_contributors["District"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Top District Contribution to Engineering Enrollment")
plt.show()
```
